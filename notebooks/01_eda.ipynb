```python
# Calculate IV for all features
def calculate_iv(data: pd.DataFrame, feature: str, target: str = "TARGET", bins: int = 10) -> float:
    """Compute Information Value (IV) for a numerical or categorical feature."""
    temp = data[[feature, target]].copy()

    if pd.api.types.is_numeric_dtype(temp[feature]) and temp[feature].nunique() > bins:
        temp["bin"] = pd.qcut(temp[feature], q=bins, duplicates="drop").astype(str)
    else:
        temp["bin"] = temp[feature].astype(str)

    temp["bin"] = temp["bin"].fillna("Missing")
    grouped = temp.groupby("bin")[target].agg(["count", "sum"])
    grouped.columns = ["total", "bad"]
    grouped["good"] = grouped["total"] - grouped["bad"]

    # Add small smoothing constant (0.5) to avoid division by zero in log
    dist_good = (grouped["good"] + 0.5) / (grouped["good"].sum() + 0.5 * len(grouped))
    dist_bad = (grouped["bad"] + 0.5) / (grouped["bad"].sum() + 0.5 * len(grouped))

    woe = np.log(dist_good / dist_bad)
    iv = ((dist_good - dist_bad) * woe).sum()
    return float(iv)


exclude_cols = {"SK_ID_CURR", "TARGET", "AGE_YEARS"}
feature_cols = [c for c in df.columns if c not in exclude_cols]

iv_results = (
    pd.DataFrame({
        "feature": feature_cols,
        "iv": [calculate_iv(df, col) for col in feature_cols],
    })
    .sort_values("iv", ascending=False)
    .reset_index(drop=True)
)

iv_results.head(20)
```

## Key EDA Findings & Modeling Decisions

1. **Class Imbalance & Evaluation Strategy:**
   - Default rate is **8.07%** (~11.4:1 imbalance).
   - Evaluation will use **Stratified 5-Fold CV** tracked via **ROC-AUC**, **Gini** ($2 \cdot \text{AUC} - 1$), and **PR-AUC**.

2. **External Bureau Scores Synergy:**
   - `EXT_SOURCE_1`, `2`, and `3` are the strongest individual predictors (IV = `0.15`, `0.31`, `0.33`) with low mutual correlation (`0.11–0.21`).
   - Combining them into `EXT_SOURCE_MEAN` nearly doubles predictive power to **IV = 0.608**.
   - *Action:* Engineer full aggregations (`mean`, `min`, `max`, `std`, `prod`, and `missing_cnt`).

3. **Multicollinearity & Domain Ratios:**
   - `AMT_CREDIT` and `AMT_GOODS_PRICE` are `0.99` correlated; transforming them into `CREDIT_TERM` (IV = `0.093`) and `CREDIT_GOODS_RATIO` (IV = `0.076`) extracts stronger signal than raw loan amounts.
   - *Action:* Add credit affordability ratios (`CREDIT_TERM`, `CREDIT_GOODS_RATIO`, `GOODS_PRICE_DIFF`, `ANNUITY_INCOME_RATIO`, `CREDIT_INCOME_RATIO`).

4. **Behavioral Stability, Geo-Risk & 2D Interactions:**
   - Strong monotonic risk decrease with longer employment tenure (`11.1%` for <3 yrs → `4.2%` for 20+ yrs), phone stability (`9.9%` → `5.4%`), and ID document age (`10.2%` → `6.4%`).
   - `REGION_RATING_CLIENT_W_CITY` separates risk by 2.4x (`4.8%` vs. `11.4%`), `ADDRESS_MISMATCH_CNT` rises from `7.3%` (0 mismatches) to `10.3–12.3%` (2–3 mismatches), and 30-DPD defaults in the social circle push risk from `~7.8%` to `~13.0%`.
   - The 2D risk matrix (`Age × EXT_SOURCE_MEAN`) achieves a **10x risk separation** (`21.4%` in Age 30–40 / Q1 vs. `2.1%` in Age 60–70 / Q5).
   - *Action:* Engineer `ADDRESS_MISMATCH_CNT`, `PHONE_TO_AGE_RATIO`, `ID_PUBLISH_TO_AGE_RATIO`, and `SOCIAL_CIRCLE_DEF_RATIO`.

5. **Informative Missingness & Near-Constant Features:**
   - Missing blocks carry clear risk signals ("thin-file" effect): missing housing data (`9.15%` vs. `6.96%` default) and missing bureau inquiries (`10.34%` vs. `7.72%`), whereas missing `OCCUPATION_TYPE` indicates lower risk (`6.51%` vs. `8.79%`, largely pensioners).
   - 16 features are near-constant (`>99.5%` single value, mostly `FLAG_DOCUMENT_*`).
   - *Action:* Preserve `NaN` values for native GBDT handling, add `TOTAL_MISSING_CNT`, and aggregate document flags into `DOCUMENTS_SUBMITTED_CNT`.

6. **Data Hygiene & Outliers:**
   - `DAYS_EMPLOYED == 365243` is the only positive anomaly across all `DAYS_*` columns. String placeholders `'XNA'` (`CODE_GENDER`) and `'Unknown'` (`NAME_FAMILY_STATUS`) appear in tiny counts.
   - Extreme numerical outliers (`AMT_INCOME_TOTAL = 117M`, `OBS_30_CNT_SOCIAL_CIRCLE = 348`, `CNT_CHILDREN = 19`) are noted; tree-based models (LightGBM/CatBoost) handle them natively.
   - *Action:* Flag `DAYS_EMPLOYED_ANOM` and replace `365243`, `'XNA'`, and `'Unknown'` with `NaN`.
