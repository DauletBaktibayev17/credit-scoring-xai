```python
# Calculate IV for all features
def calculate_iv(data: pd.DataFrame, feature: str, target: str = "TARGET", bins: int = 10) -> float:
    """Compute Information Value (IV) for a numerical or categorical feature."""
    temp = data[[feature, target]].copy()

    if pd.api.types.is_numeric_dtype(temp[feature]) and temp[feature].nunique() > bins:
        temp["bin"] = pd.qcut(temp[feature], q=bins, duplicates="drop").astype(str)
    else:
        temp["bin"] = temp[feature].astype(str)

    temp["bin"] = temp["bin"].fillna("Missing")
    grouped = temp.groupby("bin")[target].agg(["count", "sum"])
    grouped.columns = ["total", "bad"]
    grouped["good"] = grouped["total"] - grouped["bad"]

    # Add small smoothing constant (0.5) to avoid division by zero in log
    dist_good = (grouped["good"] + 0.5) / (grouped["good"].sum() + 0.5 * len(grouped))
    dist_bad = (grouped["bad"] + 0.5) / (grouped["bad"].sum() + 0.5 * len(grouped))

    woe = np.log(dist_good / dist_bad)
    iv = ((dist_good - dist_bad) * woe).sum()
    return float(iv)


exclude_cols = {"SK_ID_CURR", "TARGET", "AGE_YEARS"}
feature_cols = [c for c in df.columns if c not in exclude_cols]

iv_results = (
    pd.DataFrame({
        "feature": feature_cols,
        "iv": [calculate_iv(df, col) for col in feature_cols],
    })
    .sort_values("iv", ascending=False)
    .reset_index(drop=True)
)

iv_results.head(20)
```

## Key EDA Findings & Modeling Decisions

1. **Class Imbalance:** Default rate is **8.07%** (~11.4:1). Evaluation will use **Stratified 5-Fold CV** with **ROC-AUC**, **Gini**, and **PR-AUC**.
2. **External Scores Synergy:** `EXT_SOURCE_1/2/3` have low mutual correlation (0.11–0.21), so combining them into `EXT_SOURCE_MEAN` boosts Information Value to **0.608** (vs. 0.329 max individually). We will engineer full aggregations (mean, min, max, std, product).
3. **Multicollinearity & Domain Ratios:** `AMT_CREDIT` and `AMT_GOODS_PRICE` are 0.99 correlated; transforming them into `CREDIT_TERM` (IV=0.093) and `CREDIT_GOODS_RATIO` (IV=0.076) extracts strong signal.
4. **Categorical Risk Profiles:** Clear monotonic risk decrease by education level (10.9% Lower secondary → 5.4% Higher education). `Working` (9.6%) defaults nearly twice as often as `Pensioner` (5.4%). Rare income categories (`n < 30`) will be handled with regularization.
5. **Data Hygiene:** Flag and replace `DAYS_EMPLOYED == 365243` and `'XNA'` placeholders with `NaN`. Extreme numerical outliers (`AMT_INCOME_TOTAL = 117M`, `OBS_30_CNT_SOCIAL_CIRCLE = 348`) are noted; tree-based models are robust to them natively.
